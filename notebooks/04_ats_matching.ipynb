```python
import sys
sys.path.append("../src")

from text_extraction import extract_text_from_pdf
from skill_extraction import load_skills, extract_skills
from ats_matching import (
    load_all_job_descriptions,
    analyze_resume,
    rank_jobs
)
```

```python
pdf_path = "../data/resumes/sample_resume_2_Data_Analyst.pdf"
skills_path = "../data/skills.csv"
jd_folder = "../data/job_descriptions"
```

```python
job_descriptions = load_all_job_descriptions(jd_folder)
print(job_descriptions.keys())
```

```python
resume_text = extract_text_from_pdf(pdf_path)

skills = load_skills(skills_path)
```

```python
resume_text = extract_text_from_pdf(pdf_path)

skills = load_skills(skills_path)

resume_skills = extract_skills(resume_text, skills)

print(resume_skills)
```

```python
ats_results = analyze_resume(
    resume_skills,
    job_descriptions,
    skills
)
```

```python
for role, score in ats_results.items():
    print(role, ":", score)
```

```python
sorted_results = rank_jobs(ats_results)
```

```python
print("=" * 60)
print("Top Matching Job Roles")
print("=" * 60)

for i, (job, result) in enumerate(sorted_results, start=1):

    role = job.replace(".txt", "")
    role = role.replace("_", " ").title()

    print(f"{i}. {role:<35} {result['score']:.2f}%")
```

```python
best_role, best_result = sorted_results[0]

best_role = best_role.replace(".txt", "")
best_role = best_role.replace("_", " ").title()

print("\n" + "=" * 60)
print("BEST MATCH")
print("=" * 60)

print(f"Role : {best_role}")
print(f"ATS Score : {best_result['score']:.2f}%")
```

```python
print("\nMatched Skills")
print("-" * 60)

for skill in best_result["matched"]:
    print("✓", skill)
```

```python
print("\nMissing Skills")
print("-" * 60)

for skill in best_result["missing"]:
    print("✗", skill)
```
