```python
selected_role = "Data Analyst"

result = ats_matching.analyze_resume_for_role(
    resume_text=resume_text,
    selected_role=selected_role,
    knowledge_base_dir=KNOWLEDGE_BASE_DIR
)

print("Role:", result["role"])

print(
    "Overall Compatibility:",
    result["compatibility_score"],
    "%"
)

for category, details in result["categories"].items():

    print(f"\n{'=' * 50}")
    print(category.replace("_", " ").title())

    print("Score:", details["score"], "%")

    print("Matched Requirements:")
    for item in details["matched"]:
        print("  +", item)

    print("Missing Requirements:")
    for item in details["missing"]:
        print("  -", item)

    print("Supporting Evidence:")
    for item in details["matched_evidence"]:
        print("  ", item["requirement"], ":", item["evidence"])
```
